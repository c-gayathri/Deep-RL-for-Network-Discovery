```python
import networkx as nx
import itertools
import random
```

```python
ratios = [0.4, 0.4, 0.2]
n = 700
p_in = 0.005
p_out = 0.001

communities = {}
edges = set()

nodes = list(range(n))
start = 0
com = 0
for r in ratios:
    end = int(r*n) + start
    if end > n:
        end = n
    print()
    print('start:')
    print(start)
    print('end:')
    print(end)
    comNodes = dict.fromkeys(nodes[start:end], com)
    communities.update(comNodes)
    start = end
    com += 1
    
print(communities)
    
for combination in itertools.combinations(nodes, 2):
    x, y = combination
    print(x)
    print(y)
    if communities[x] == communities[y]:
        if random.random() < p_in:
            edges.add(combination)
    else:
        if random.random() < p_out:
            edges.add(combination)

G = nx.Graph()
G.add_nodes_from(nodes)
G.add_edges_from(edges)
```
